````python
import torch
````

* A torch.Tensor is a multi-dimentional matrix conatining elements of a single datatype.
* Similar to NumPy arrays, but with better compatibility with GPU's.
* Default datatype is float32.
* More suitable for deep learning than NumPy arrays.

## Creating tensors

### Creating a tensor using data

````python
data = [[1,2,3], [4,5,6]]
my_tensor = torch.tensor(data)
````

````python
print(my_tensor)
````

### Creating from a desired shape

````python
# For when we do not know the values of the tnesor, and only the shape of it

shape = (3,5)
````

````python
ones = torch.ones(shape)  # Will create a tensor with the desired shape, all values initialised to 1

zeros = torch.zeros(shape)  # Will create a tensor with the desired shape but all values initialised to 0

random = torch.randn(shape)  # Will create the tensor with the desired shape, but with random values
````

````python
print(f"Ones tensor: \n {ones} \n")

print(f"Zeros tensor: \n {zeros} \n")

print(f"Random tensor: \n {random} \n")
````

### Using a template

````python
# Using a template tensor to create a new tensor

template = torch.tensor([[1,2,3],[4,5,6]])
````

````python
random_tensor = torch.randn_like(template, dtype=torch.float)
````

````python
print(f"Random tensor from template: \n {random_tensor} \n")
````

## Autograd in PyTorch

PyTorch consist of an **autograd**(automatic differentiation) that defines which tensors(objects) are **learnable parameters** and which are not.

Activated using:

```
requires_grad = true
```
> This defines that the tensor is a learnable parameter, and autograd should **track *every single operation*** happening to it.


````python
normal_tensor = torch.tensor([[1.,2.,3.],[4.,5.,6.]])

param_tensor = torch.tensor([[1.,2.], [3.,4.]], requires_grad=True)

print(f"Normal Tensor requires_grad: {normal_tensor.requires_grad} \n")

print(f"Parameter Tensor requires_grad: {param_tensor.requires_grad}")
````

## Building a graph

````python
a = torch.tensor([2.0], requires_grad = True)
b = torch.tensor([3.0], requires_grad = True)
x = torch.tensor([4.0], requires_grad = True)
````

````python
y = a + b

z = x * y
````

````python
print(f"result z: {z}")
````

````python
print(f"z grad_fn: {z.grad_fn} \n")

print(f"y grad_fn: {y.grad_fn} \n")

print(f"a grad_fn: {a.grad_fn}")
````

The above **grad_fn** is the computational graph that PyTorch keeps track of.

It tells how the current tensor came to be or from where it originated.

Then by using *loss.backward()*, we can track these values and calculate the **gradients**.
